```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')
```

```python
df=pd.read_csv("Automobile_data.csv")
```

```python
df.head()
```

```python
df["normalized-losses"].replace("?",np.nan,inplace=True)
```

```python
df.isnull().sum()
```

```python
df["normalized-losses"] = pd.to_numeric(
    df["normalized-losses"],
    errors="coerce"
)
```

```python
m =df["normalized-losses"].mean()
print("Mean of normalized-losses : ",m)
df["normalized-losses"].fillna(m,inplace=True)
```

```python
df.isnull().sum()
```

```python
"?" in df["horsepower"].values
(df["horsepower"] == "?").sum()
df[df["horsepower"] == "?"]
df["horsepower"] = df["horsepower"].replace("?", np.nan)
df["horsepower"] = df["horsepower"].astype("float")
df.isnull().sum()
```

```python
m = df["horsepower"].mean()
print("Mean of horsepower:", m)
df["horsepower"] = df["horsepower"].fillna(m, inplace = True)
```

```python
df.isnull().sum()
```

```python
df_num=df.select_dtypes(["int64","float64"])#hold to int and float type data
df_cat=df.select_dtypes(object)#hold object type data
```

```python
from sklearn.preprocessing import LabelEncoder
for col in df_cat :
    le=LabelEncoder() #create the object of LabelEncoder class
    df_cat[col]=le.fit_transform(df_cat[col])
```

```python
df_cat.head()
```

```python
df_new=pd.concat([df_num,df_cat],axis=1) #axis=1 column wise
df_new.head()
```

```python

#select independent means input variables(X) and target means dependent
#variable(Y)
X=df_new.drop("price",axis=1) #apart from price, all columns hold in X variable
Y=df_new["price"] #output variable
```

```python
#train test split 70%-30%
from sklearn.model_selection import train_test_split
X_train,X_test,Y_train,Y_test=train_test_split(X,Y,test_size=0.3,
                                              random_state=1)
```

```python
X_train.shape,X_test.shape
```

```python
#Apply StandardScaler on X_train and X_test
from sklearn.preprocessing import StandardScaler
#create the object of StandardScaler class
ss=StandardScaler()
X_train=ss.fit_transform(X_train)
X_test=ss.transform(X_test)
```

```python
X_train
```

```python
from sklearn.impute import SimpleImputer

imputer = SimpleImputer(strategy='median')

X_train = imputer.fit_transform(X_train)
X_test = imputer.transform(X_test)
```

```python
#PERFORM MODEL WITH LINEARREGRESSION
from sklearn.linear_model import LinearRegression
#create object of LinearRegression class
lr=LinearRegression()
#train the model with 70% data
lr.fit(X_train,Y_train)
```

```python
y_pred = lr.predict(X_test)
```

```python
y_pred[:10]
```

```python
comparison = pd.DataFrame({ "Actual": Y_test, "Predicted": y_pred })
comparison.head(10)
```

```python
from sklearn.metrics import mean_absolute_error
```

```python
mae = mean_absolute_error(Y_test, y_pred)
print("Mean Absolute Error:", mae)
```

```python
from sklearn.metrics import mean_squared_error
```

```python
mse = mean_squared_error(Y_test, y_pred)
print("Mean Squared Error:", mse)
```

```python
from sklearn.metrics import r2_score
```

```python
r2 = r2_score(Y_test, y_pred)
print("R² Score:", r2)
```

Task 2: Apply Lasso Regression


```python
from sklearn.linear_model import Lasso
```

```python
for i in range(200,1001,50):
    l1=Lasso(i) 
    l1.fit(X_train,Y_train)
    print("Value of Lambda : ",i)
   
    print("Training Score : ",l1.score(X_train,Y_train))
    
    print("Testing Score : ",l1.score(X_test,Y_test))
```

```python
l1.coef_
```

```python
df1['Slope_Lasso']=l1.coef_
df1
```

```python
lasso_model = Lasso(alpha= 200)
```

```python
lasso_model.fit(X_train, Y_train)
```

```python
lasso_pred = lasso_model.predict(X_test)
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score 
print("MAE :", mean_absolute_error(Y_test, lasso_pred)) 
print("MSE :", mean_squared_error(Y_test, lasso_pred))
print("R² Score :", r2_score(Y_test, lasso_pred))
```

Task 3: Apply Ridge Regression

```python
from sklearn.linear_model import Ridge
```

```python
for i in range(0,51):
    l2=Ridge(i)
    l2.fit(X_train,Y_train)
    print("Value of Lambda : ",i)
    
    print("Training Score : ",l2.score(X_train,Y_train))
    
    print("Testing Score : ",l2.score(X_test,Y_test))
```

```python
l2.coef_
```

```python
df1['Slope_Ridge']=l2.coef_
df1
```

```python
ridge_model = Ridge(alpha= 12)
```

```python
ridge_model.fit(X_train, Y_train)
```

```python
ridge_pred = ridge_model.predict(X_test)
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
print("MAE :", mean_absolute_error(Y_test, ridge_pred))
print("MSE :", mean_squared_error(Y_test, ridge_pred))
print("R² Score :", r2_score(Y_test, ridge_pred))
```
